# Exoplanet Model Lab
Explore a catalog and evaluate a mass regressor with the same training code as the app. The default dataset is synthetic.

```python
from pathlib import Path
import pandas as pd
from exoplanet import default_features, demo_data, feature_columns, predict, read_csv, train

csv_path = None  # Set to Path("planets.csv") to use a real catalog.
target = "pl_bmassj"
task = "regression"
df, truncated = read_csv(csv_path) if csv_path else (demo_data(), False)
print(f"{len(df):,} rows" + (" (truncated)" if truncated else ""))
df.head()
```

```python
features = default_features(feature_columns(df, target))
print("Features:", features)
df[features + [target]].isna().sum().rename("Missing values")
```

## Evaluate on held-out rows
The training split supplies the medians used to fill missing features. Missing targets are excluded. Check the README for limits of random holdout evaluation.

```python
artifact, report = train(df, features, target, task)
print(f"Train: {report['train_rows']:,}; holdout: {report['test_rows']:,}")
pd.Series(report["metrics"], name="Holdout score")
```

```python
report["importance"].rename("Feature importance")
```

## Predict for a new example
Replace the example values with measurements in the same units as the training data.

```python
example = pd.DataFrame([artifact["defaults"]])
predict(artifact, example)
```
