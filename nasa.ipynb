## Cleaning validation checklist
Run the following cell to get a concise report on: nulls, duplicates, columns containing ± (uncertainties), object-columns that are mostly numeric, leading/trailing whitespace, and ID-like columns with special characters.
The cell will show sample problematic rows and a short recommendation list.

```python
for c in df.select_dtypes(include=[np.number]).columns:
    if df[c].isnull().any():
        df[c].fillna(df[c].median(), inplace=True)

for c in df.select_dtypes(include=['object']).columns:
    if df[c].isnull().any():
        try:
            df[c].fillna(df[c].mode(dropna=True)[0], inplace=True)
        except:
            pass
```

```python
# Drop object columns with too many unique values (e.g. >100)
max_unique = 100
for col in df.select_dtypes(include='object').columns:
    if df[col].nunique() > max_unique:
        print(f"Dropping column '{col}' due to too many categories ({df[col].nunique()})")
        df.drop(columns=col, inplace=True)
```

```python
for col in df.select_dtypes(include=['number']).columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    try:
        df[col] = df[col].astype('Int64')
    except TypeError:
        pass

for col in df.select_dtypes(include=['number']).columns:
    df[col] = df[col].astype(float)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, accuracy_score, classification_report
import numpy as np
import pandas as pd

def train_model_and_predict(df, sel=None, tar='pl_bmassj', new_df=None, n_estimators=100, random_state=42):
    if sel is None:
        sel = [c for c in ['pl_orbper','sy_jmag','sy_hmag','sy_kmag','disc_year'] if c in df.columns]
        if not sel:
            sel = list(df.select_dtypes(include=[np.number]).columns[:5])
    print('Using features:', sel)

    X = df[sel].copy()
    for c in X.columns:
        X[c] = pd.to_numeric(X[c], errors='coerce')
    X = X.fillna(X.median())

    if tar and (tar in df.columns):
        y = pd.to_numeric(df[tar], errors='coerce')
        if y.isnull().any():
            y = y.fillna(y.median())
        print('Training regressor for target:', tar)
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=random_state)
        model = RandomForestRegressor(n_estimators=n_estimators, random_state=random_state)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        print('MSE:', mean_squared_error(y_test, preds))
        print('MAE:', mean_absolute_error(y_test, preds))
        print('R2:', r2_score(y_test, preds))

        if new_df is not None:
            new_X = new_df[sel].copy()
            for c in new_X.columns:
                new_X[c] = pd.to_numeric(new_X[c], errors='coerce')
            new_X = new_X.fillna(X.median())
            new_preds = model.predict(new_X)
            print('Predictions for new rows (regression target):')
            print(new_preds)
        return model

    indicator_cols = [c for c in ['pl_name','pl_bmassj','pl_orbper','pl_rade','pl_radj'] if c in df.columns]
    if not indicator_cols:
        print('No indicator columns found to construct is_planet label; cannot train classifier.')
        return None

    y = df[indicator_cols].notnull().any(axis=1).astype(int)
    print('Built is_planet label from columns:', indicator_cols)
    if y.nunique() < 2:
        print('Not enough label variation to train a classifier (all examples same class).')
        return None

    stratify_arg = y if (y.value_counts().min() >= 2) else None
    if stratify_arg is None:
        print('Proceeding without stratify (some classes have <2 samples)')

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=random_state, stratify=stratify_arg)
    clf = RandomForestClassifier(n_estimators=n_estimators, random_state=random_state)
    clf.fit(X_train, y_train)
    preds = clf.predict(X_test)
    print('Accuracy:', accuracy_score(y_test, preds))
    print('Classification report:')
    print(classification_report(y_test, preds))

    if new_df is not None:
        new_X = new_df[sel].copy()
        for c in new_X.columns:
            new_X[c] = pd.to_numeric(new_X[c], errors='coerce')
        new_X = new_X.fillna(X.median())
        new_preds = clf.predict(new_X)
        print('Predictions for new rows (1=planet,0=not):', new_preds)

    return clf


```
